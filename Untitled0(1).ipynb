## Section 3: Load Whisper Model

Now, we'll load the pretrained Whisper Small model and its corresponding processor from Hugging Face. The processor handles tokenization and feature extraction, while the model performs the actual speech-to-text transcription. We'll also check for and utilize a GPU if available for faster processing.

````python
# Determine if a GPU is available and set the device accordingly
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load Whisper processor and model
# The processor prepares the audio input for the model
processor = WhisperProcessor.from_pretrained("openai/whisper-small")
# The model performs the speech-to-text transcription
model = WhisperForConditionalGeneration.from_pretrained("openai/whisper-small").to(device)

# Configure the model to use the language specified by the user (Egyptian Arabic is not directly supported by Whisper-small,
# but we can set the task to 'transcribe' and language to 'arabic' for general Arabic transcription.
# Fine-tuning would be needed for Egyptian Arabic specifics.)
model.config.forced_decoder_ids = processor.get_decoder_prompt_ids(language="arabic", task="transcribe")

print("Whisper model and processor loaded successfully.")
````

## Section 5: Load Summarization Model

Now that we have the transcript, we'll load a pretrained sequence-to-sequence model for summarization. FLAN-T5 is a good choice for its ability to follow instructions and its relatively lightweight nature. We'll use the `google/flan-t5-small` version. We'll load both the tokenizer and the model, moving the model to the GPU if available.

````python
# Load FLAN-T5 tokenizer and model for summarization
# The tokenizer converts text into numerical IDs for the model
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
# The model performs text-to-text generation tasks like summarization
summarization_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small").to(device)

print("FLAN-T5 summarization model and tokenizer loaded successfully.")
````

## Section 6: Generate Summary, Bullet Points, and Action Items

With the transcript and the summarization model ready, we can now generate the desired outputs: a summary, bullet points, and action items. We'll construct specific prompts for the FLAN-T5 model to guide its generation for each task. Since FLAN-T5 is an instruction-tuned model, clear instructions in the prompt are crucial.

````python
def generate_text_output(text, task_prompt, model, tokenizer, device, max_length=150, num_beams=4, early_stopping=True):
    """
    Generates text output (summary, bullet points, action items) using a given model and prompt.

    Args:
        text (str): The input text (transcript).
        task_prompt (str): The instruction for the model (e.g., "Summarize:").
        model: The pretrained sequence-to-sequence model.
        tokenizer: The tokenizer for the model.
        device (str): The device ('cuda' or 'cpu') where the model is loaded.
        max_length (int): Maximum length of the generated output.
        num_beams (int): Number of beams for beam search decoding. Higher values improve quality but increase computation.
        early_stopping (bool): Whether to stop generation early if all beam hypotheses have reached an EOS token.

    Returns:
        str: The generated text output.
    """
    # Combine the task prompt with the input text
    input_text = f"{task_prompt} {text}"

    # Tokenize the input text and move to device
    inputs = tokenizer(input_text, return_tensors="pt", max_length=512, truncation=True).to(device)

    # Generate output using the model
    outputs = model.generate(
        inputs["input_ids"],
        max_length=max_length,
        num_beams=num_beams,
        early_stopping=early_stopping
    )

    # Decode the generated token IDs back to text
    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return generated_text

# Define prompts for each task
summary_prompt = "Summarize the following text:"
bullet_points_prompt = "Generate bullet points from the following text:"
action_items_prompt = "Extract any action items from the following text:"

# Generate summary
print("\nGenerating Summary...")
summary = generate_text_output(transcription, summary_prompt, summarization_model, tokenizer, device)
print("--- Summary ---")
print(summary)

# Generate bullet points
print("\nGenerating Bullet Points...")
bullet_points = generate_text_output(transcription, bullet_points_prompt, summarization_model, tokenizer, device)
print("--- Bullet Points ---")
print(bullet_points)

# Generate action items
print("\nGenerating Action Items...")
action_items = generate_text_output(transcription, action_items_prompt, summarization_model, tokenizer, device)
print("--- Action Items ---")
print(action_items)
````

## Section 7: Run the Complete Pipeline

This section consolidates all the previous steps into a single, comprehensive function or block of code. You can use this to process any new audio file from start to finish, generating a transcript, summary, bullet points, and action items. This demonstrates the complete "Fakarny" prototype pipeline.

````python
# Import standard libraries (re-import to ensure all are available)
import torch
import pandas as pd

# Import Hugging Face Transformers libraries
from transformers import WhisperProcessor, WhisperForConditionalGeneration
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Import audio processing libraries
import librosa
import soundfile as sf
from datasets import load_dataset, Audio

# Other utilities
import os
import tempfile
from pathlib import Path

def generate_text_output(text, task_prompt, model, tokenizer, device, max_length=150, num_beams=4, early_stopping=True):
    """
    Generates text output (summary, bullet points, action items) using a given model and prompt.
    """
    input_text = f"{task_prompt} {text}"
    inputs = tokenizer(input_text, return_tensors="pt", max_length=512, truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}
    outputs = model.generate(
        inputs["input_ids"],
        attention_mask=inputs.get("attention_mask"),
        max_length=max_length,
        num_beams=num_beams,
        early_stopping=early_stopping,
    )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

def run_fakarny_pipeline(audio_path, whisper_processor, whisper_model, summarization_tokenizer, summarization_model, device):
    print(f"\n--- Starting Fakarny Pipeline for {audio_path} ---")
    audio, sampling_rate = librosa.load(audio_path, sr=16000)
    print("Transcribing audio...")
    input_features = whisper_processor(audio, sampling_rate=16000, return_tensors="pt").input_features.to(device)
    predicted_ids = whisper_model.generate(input_features)
    transcript = whisper_processor.batch_decode(predicted_ids, skip_special_tokens=True)[0]
    print("\n--- Transcript ---")
    print(transcript)
    print("\nGenerating Summary, Bullet Points, and Action Items...")
    summary = generate_text_output(transcript, summary_prompt, summarization_model, summarization_tokenizer, device)
    bullet_points = generate_text_output(transcript, bullet_points_prompt, summarization_model, summarization_tokenizer, device)
    action_items = generate_text_output(transcript, action_items_prompt, summarization_model, summarization_tokenizer, device)
    print("\n--- Generated Outputs ---")
    print("**Summary:**")
    print(summary)
    print("\n**Bullet Points:**")
    print(bullet_points)
    print("\n**Action Items:**")
    print(action_items)
    return {"transcript": transcript, "summary": summary, "bullet_points": bullet_points, "action_items": action_items}

# Determine if a GPU is available and set the device accordingly
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load Whisper processor and model
processor = WhisperProcessor.from_pretrained("openai/whisper-tiny")
model = WhisperForConditionalGeneration.from_pretrained("openai/whisper-tiny").to(device)
model.config.forced_decoder_ids = processor.get_decoder_prompt_ids(language="arabic", task="transcribe")
print("Whisper model and processor loaded successfully (re-initialized).")

# Load FLAN-T5 tokenizer and model for summarization
tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
summarization_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small").to(device)
print("FLAN-T5 summarization model and tokenizer loaded successfully (re-initialized).")

# Define prompts for each task (re-define)
summary_prompt = "Summarize the following text:"
bullet_points_prompt = "Generate bullet points from the following text:"
action_items_prompt = "Extract any action items from the following text:"

# Load the local dataset labels and audio paths
workspace_root = Path.cwd()
dataset_root = workspace_root / "dataset"
audio_dir = dataset_root / "audio_files"
labels_path = dataset_root / "labels.csv"
labels_df = pd.read_csv(labels_path)
labels_df["audio_path"] = labels_df["path"].apply(lambda p: str(audio_dir / p))
labels_df = labels_df[labels_df["audio_path"].apply(os.path.exists)].reset_index(drop=True)
print(f"Loaded {len(labels_df)} labeled audio files from {dataset_root}")

# Run the pipeline on the first labeled sample
sample_row = labels_df.iloc[0]
user_audio_path = sample_row["audio_path"]
print(f"Using sample audio from: {user_audio_path}")
print(f"Label: {sample_row["label"]}")

pipeline_results = run_fakarny_pipeline(
    audio_path=user_audio_path,
    whisper_processor=processor,
    whisper_model=model,
    summarization_tokenizer=tokenizer,
    summarization_model=summarization_model,
    device=device,
)
````

### Fine-tuning Whisper for Egyptian Arabic Speech-to-Text

Pretrained Whisper models, while powerful, are primarily trained on diverse English audio. For optimal performance with Egyptian Arabic, you would need to fine-tune `openai/whisper-small` on a relevant dataset.

1.  **Dataset Collection:** Acquire or create a dataset of Egyptian Arabic speech audio paired with accurate text transcripts. This is the most crucial step. Datasets like [Common Voice Arabic](https://huggingface.co/datasets/mozilla-foundation/common_voice_16_1_ar) might have some Egyptian Arabic, but a dedicated dataset would be best.
2.  **Data Preparation:** The audio needs to be sampled at 16kHz and transcripts should be cleaned and normalized.
3.  **Fine-tuning Script:** Use the `transformers` library's `Trainer` API or write a custom PyTorch training loop. Key considerations:

    *   **Training Arguments:** Define learning rate, batch size, number of epochs, etc.
    *   **Evaluation Metrics:** Use Word Error Rate (WER) as the primary metric for ASR.

### Fine-tuning FLAN-T5 for Specific Summarization, Bullet Points, and Action Items

The current FLAN-T5 model is general-purpose. Fine-tuning it on task-specific data will yield much better results for your exact requirements.

1.  **Dataset Collection:** This is crucial. You need pairs of:
    *   `(transcript, desired_summary)`
    *   `(transcript, desired_bullet_points)`
    *   `(transcript, desired_action_items)`
    You might need to manually annotate a small dataset initially or find existing resources.
2.  **Data Preparation:** Tokenize the input and target texts using the FLAN-T5 tokenizer.
3.  **Prompt Engineering:** During fine-tuning, you would continue to use structured prompts (like "Summarize: ", "Extract Action Items: ") so the model learns to respond to these specific instructions.
4.  **Fine-tuning Script:** Similar to Whisper, use the `transformers.Trainer`:

    *   **Training Arguments:** Define parameters like learning rate, epochs, etc.
    *   **Evaluation Metrics:** Use ROUGE scores for summarization, and potentially custom metrics for bullet points/action items (e.g., precision/recall of extracted phrases).

### General Advice for Fine-tuning:

*   **Start Small:** Begin with smaller models (like `whisper-tiny` or `flan-t5-small`) and smaller datasets to establish a baseline before scaling up.
*   **Monitor Progress:** Use tools like TensorBoard (integrated with `Trainer`) to monitor training loss and evaluation metrics.
*   **Hardware:** Access to a powerful GPU (e.g., V100, A100) or multiple GPUs will significantly speed up fine-tuning.
*   **Transfer Learning:** You are already using transfer learning by starting with pretrained models. Fine-tuning simply adapts these powerful models to your specific domain and task with less data than training from scratch.
